### 4. Відображення (map та apply)
Використовуємо метод `map()` для перетворення одного стовпця, а `apply()` — для обробки рядків датафрейму.

Центруємо світові продажі відносно середнього значення за допомогою `map()`.

```python
games_sales_mean = games.Global_Sales.mean()
games.Global_Sales.map(lambda s: s - games_sales_mean)
```

Створюємо функцію для рядка датафрейму: віднімаємо середнє від `Global_Sales` і додаємо новий стовпець `Revenue_Estimate` (орієнтовна виручка у млн $ при середній ціні $50 за гру).

```python
def remean_sales(row):
    row.Global_Sales = row.Global_Sales - games_sales_mean
    row['Revenue_Estimate'] = (row.Global_Sales + games_sales_mean) * 50
    return row

games.apply(remean_sales, axis='columns')
```

Перевіряємо, що сам датафрейм `games` не змінився (методи `map` та `apply` повертають нові об'єкти, а не модифікують оригінал).

```python
games.head()
```

### 5. Вбудовані векторизовані операції
Замінюємо метод `map()` прямою векторизованою операцією віднімання середнього значення.

```python
games_sales_mean = games.Global_Sales.mean()
games.Global_Sales - games_sales_mean
```

Порівнюємо швидкість роботи методу `map()` та вбудованої векторизованої операції за допомогою `%timeit`.

```python
# Замір часу для map
%timeit games.Global_Sales.map(lambda s: s - games_sales_mean)
```

```python
# Замір часу для векторизованої операції
%timeit games.Global_Sales - games_sales_mean
```

Виконуємо векторизовані операції між двома стовпцями:
1. Конкатенація двох текстових стовпців (`Name` та `Platform`);
2. Додавання двох числових стовпців (`NA_Sales` та `EU_Sales`).

```python
# Об'єднання назви гри та платформи
games.Name + " (" + games.Platform + ")" 
```

```python
# Сумарні продажі в Північній Америці та Європі
games.NA_Sales + games.EU_Sales
```

### 6. Візуалізація результатів для звіту

```python
import os
os.makedirs('figures', exist_ok=True)

# 1. Розподіл кількості ігор за жанрами
plt.figure(figsize=(10, 5))
sns.countplot(data=games, y='Genre', order=games['Genre'].value_counts().index, hue='Genre', legend=False, palette='Blues_r')
plt.title('Розподіл відеоігор за жанрами')
plt.xlabel('Кількість ігор')
plt.ylabel('Жанр')
plt.grid(True, alpha=0.3)
plt.savefig('figures/genres.png', bbox_inches='tight')
plt.show()
```

```python
# 2. Топ-10 найбільш продаваних ігор у світі
top10 = games.sort_values(by='Global_Sales', ascending=False).head(10)

plt.figure(figsize=(10, 5))
plt.barh(top10['Name'] + " (" + top10['Platform'] + ")", top10['Global_Sales'], color='steelblue')
plt.gca().invert_yaxis()
plt.title('Топ-10 найбільш продаваних відеоігор (світовий тираж)')
plt.xlabel('Глобальні продажі (млн копій)')
plt.ylabel('Гра')
plt.grid(True, alpha=0.3)
plt.savefig('figures/top10.png', bbox_inches='tight')
plt.show()
```
